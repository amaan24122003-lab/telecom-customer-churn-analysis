```python
# Which contract type has the highest churn rate?

customer_churn['Churn_rate'].sort_values(ascending = False).round(2)
```

# C. Revenue and pricing analysis
- Do customers with higher monthly charges churn more?
- How much recurring revenue is associated with churned customers?
- Are certain payment methods associated with higher churn?

```python
df[['MonthlyCharges', 'TotalCharges']].dtypes
```

```python
# 1.Do customers with higher monthly charges churn more?
# find the average of churn customer

monthly_churn = df.groupby('Churn')['MonthlyCharges'].mean()
monthly_churn.round(2)
```

```python
# 2.Why calculate the median too? The average can be influenced by unusually high charges. 
# The median helps you compare the typical customer in each group.

df.groupby('Churn')['MonthlyCharges'].median().round(2)
```

```python
# visualization for --> Churn vs Monthly_Charges

sns.boxplot(
    data = df,
    x = 'Churn',
    y = 'MonthlyCharges',
    hue = 'Churn',
    width = 0.5
)

plt.title('Monthly Charges by Customer Churn')
plt.xlabel('Churn (No / Yes)')
plt.ylabel('Monthly Charges')
plt.show()
```

```python
# 3.How much recurring revenue is associated with churned customers?
# Step 1: Calculate monthly revenue associated with churn
churned_customers = df[df['Churn'] == 'Yes']
monthly_revenue_lost = churned_customers['MonthlyCharges'].sum()
print("Monthly recurring revenue associated with churn:", monthly_revenue_lost.round(2))
```

```python
# Step 2: Calculate the revenue associated with all customers

total_monthly_revenue = df['MonthlyCharges'].sum()

percentage = (
    monthly_revenue_lost / total_monthly_revenue
)*100
print("Total monthly charges:", round(total_monthly_revenue, 2))
print("Churn-associated revenue (%):",
      f"{round(percentage, 2)}%")
```

```python
# Step 3: Compare revenue from churned and retained customers

revenue_by_churn = df.groupby('Churn').agg(
    Customer_Count=('Churn', 'size'),
    Monthly_Revenue=('MonthlyCharges', 'sum'),
    Average_Monthly_Charge=('MonthlyCharges', 'mean')
)

revenue_by_churn.round(2)
```

```python
# step 4: visualization

revenue_data = revenue_by_churn.reset_index()

sns.barplot(
    data = revenue_data,
    x = 'Churn',
    y = 'Monthly_Revenue',
    hue = 'Churn',
    width = 0.4
)

plt.title('Monthly Revenue by Churn Status')
plt.xlabel('Churn (No / Yes)')
plt.ylabel('Total Monthly Charges')
plt.show()
```

# D. Service and customer experience
- Do customers without tech support churn more?
- Does internet service type relate to churn?
- Do customers without online security or backup services leave more often?

```python
# 1.Do customers without tech support churn more?
# Step 1: Calculate churn rate

tech_churn = df.groupby('TechSupport').agg(
    Total_Customers = ('Churn', 'size'),
    Churned_Customers = ('Churn', lambda x: (x == 'Yes').sum())
)

tech_churn['Churn_Rate'] = (
    tech_churn['Churned_Customers']
    / tech_churn['Total_Customers']
) * 100

tech_churn.round(2)
```

```python
# step 2: visualization

sns.barplot(
    data = tech_churn.reset_index(),
    x = 'TechSupport',
    y = 'Churn_Rate',
    hue = 'TechSupport',
    width = 0.5
)

plt.title('Churn Rate by Tech Support')
plt.xlabel('Tech Support')
plt.ylabel('Churn Rate (%)')

plt.show()
```

```python
# 2.Does internet service type relate to churn?
# Step 1: Calculate churn rate by internet service

internet_churn = df.groupby('InternetService').agg(
    Total_Customers=('Churn', 'size'),
    Churned_Customers=('Churn', lambda x: (x == 'Yes').sum())
)

internet_churn['Churn_Rate'] = (
    internet_churn['Churned_Customers']
    / internet_churn['Total_Customers']
) * 100

internet = internet_churn.sort_values(
    by='Churn_Rate',
    ascending=False
).round(2)

int1 = internet.reset_index()
int1
```

```python
# Step 2: Create a bar chart

sns.barplot(
    data = int1,
    x = 'InternetService',
    y = 'Churn_Rate',
    hue = 'InternetService',
    width = 0.4
)

plt.title('Churn Rate by Internet Service Type')
plt.xlabel('Internet Service')
plt.ylabel('Churn Rate (%)')
plt.show()
```

```python
# 3.Do customers without online security or backup services leave more often?
# Step 1: Calculate churn rate for both services

def funct_service(columns):
    result = df.groupby(columns).agg(
        Total_Customers=('Churn', 'size'),
        Churned_Customers=('Churn',lambda x: (x == 'Yes').sum())
    )
    
    result['Churn_Rate'] = (result['Churned_Customers'] / result['Total_Customers'])*100

    return result.round(2)
```

```python
security_churn = funct_service('OnlineSecurity')
s1 = security_churn.reset_index()
s1
```

```python
backup_churn = funct_service('OnlineBackup')
b1 = backup_churn.reset_index()
b1
```

```python
# Step 2: Visualize online security

sns.barplot(
    data = s1,
    x = 'OnlineSecurity',
    y = 'Churn_Rate',
    width = 0.4
)

plt.title('Churn Rate by Online Security')
plt.xlabel('Online Security')
plt.ylabel('Churn Rate (%)')
plt.show()
```

```python
# Step 3: Visualize online backup

sns.barplot(
    data=backup_churn.reset_index(),
    x='OnlineBackup',
    y='Churn_Rate',
    hue = 'Churn_Rate',
    palette = 'Set2',
    width = 0.5
)

plt.title('Churn Rate by Online Backup')
plt.xlabel('Online Backup')
plt.ylabel('Churn Rate (%)')
plt.show()
```

# E. Customer tenure analysis
- Are new customers more likely to leave?
- How does churn change as customers stay longer?
- Which customer groups should receive retention offers first?

```python
# 1.Are new customers more likely to leave?
# Step 1:Check the tenure column

df['tenure'].describe()
```

```python
# Step 2: Divide customers into tenure groups

bins = [0, 12, 24, 48, 72]
labels = ['0-12 month', '12-24 month', '24-48 month', '48-72 month']

df['Tenure_Group'] = pd.cut(
    df['tenure'],
    bins = bins,
    labels = labels,
    include_lowest = True
)
```

```python
# Step 3: Calculate churn rate for each group

tenure_churn = df.groupby('Tenure_Group', observed = True).agg(
    Total_customers = ('Churn', 'size'),
    Churned_customers =(
        'Churn',
        lambda x: (x == 'Yes').sum()
    )
)

tenure_churn['Churn_Rate'] = (
    tenure_churn['Total_customers'] / 
    tenure_churn['Churned_customers']
)*100

ten_chu = tenure_churn.round(2)
ten_chu
```

```python
# Step 4: Visualize the results

sns.barplot(
    data = ten_chu.reset_index(),
    x = 'Tenure_Group',
    y = 'Churn_Rate',
    width = 0.5
)

plt.title('Churn Rate by Customer Tenure')
plt.xlabel('Customer Tenure')
plt.ylabel('Churn Rate (%)')
plt.show()
```

```python
# 2. How does churn change as customers stay longer?
# Step 1: Calculate churn rate for each tenure month

monthly_tenure_churn = df.groupby('tenure').agg(
    Total_Customers=('Churn', 'size'),
    Churned_Customers=(
        'Churn',
        lambda x: (x == 'Yes').sum()
    )
)

monthly_tenure_churn['Churn_Rate'] = (
    monthly_tenure_churn['Churned_Customers']
    / monthly_tenure_churn['Total_Customers']
) * 100

monthly_tenure_churn.head(10).round(2)
```

```python
# Step 2: Create a line chart

sns.lineplot(
    data=monthly_tenure_churn.reset_index(),
    x='tenure',
    y='Churn_Rate',
    marker='o',
    color = 'red'
)

plt.title('Customer Churn Rate Across Tenure Months')
plt.xlabel('Tenure (Months)')
plt.ylabel('Churn Rate (%)')
plt.grid(alpha=0.3)
plt.show()
```

```python
# 3. Which customer groups should receive retention offers first?
# Step 1: Analyze tenure and contract together

segment_churn = df.groupby(
    ['Tenure_Group', 'Contract'],
    observed=True
).agg(
    Total_Customers=('Churn', 'size'),
    Churned_Customers=(
        'Churn',
        lambda x: (x == 'Yes').sum()
    )
)

segment_churn['Churn_Rate'] = (
    segment_churn['Churned_Customers']
    / segment_churn['Total_Customers']
) * 100

segment_churn = segment_churn.reset_index()

segment_churn.sort_values(
    'Churn_Rate',
    ascending=False
).round(2)
```

```python
# Step 2: Visualize the customer segments

heatmap_data = segment_churn.pivot(
    index='Tenure_Group',
    columns='Contract',
    values='Churn_Rate'
)

plt.figure(figsize=(9, 5))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt='.1f',
    cmap='Reds',
    linewidths=0.5
)

plt.title('Churn Rate by Tenure and Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Customer Tenure')
plt.show()
```
